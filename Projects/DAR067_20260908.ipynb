```python
%sql
WITH batch0 AS (
    SELECT DISTINCT PERSON_ID 
    FROM 5_projects.dar067.imaging_transfer_20260807
),
batch_assign AS (
    SELECT 
    p.PERSON_ID,
    1 + (pmod(abs(hash(p.PERSON_ID)), 3)) AS batch_number
    FROM 5_projects.dar067.rde_patient_demographics AS p
    LEFT JOIN batch0
    ON p.PERSON_ID = batch0.PERSON_ID
    WHERE batch0.PERSON_ID IS NULL
)
SELECT * FROM batch_assign

```

```python
%sql
WITH batch0 AS (
    SELECT DISTINCT PERSON_ID 
    FROM 5_projects.dar067.imaging_transfer_20260807
),
batch_assign AS (
    SELECT 
    p.PERSON_ID,
    1 + (pmod(abs(hash(p.PERSON_ID)), 3)) AS batch_number
    FROM 5_projects.dar067.rde_patient_demographics AS p
    LEFT JOIN batch0
    ON p.PERSON_ID = batch0.PERSON_ID
    WHERE batch0.PERSON_ID IS NULL
)
SELECT DISTINCT im.AccessionNbr AS AccessionNumber FROM 4_prod.pacs.imaging_metadata AS im
INNER JOIN (SELECT DISTINCT PERSON_ID FROM batch_assign WHERE batch_number = 3) AS ba
ON ba.PERSON_ID = im.PERSONID
WHERE ExamCode IN ('CLIVE', 'CLIVEC', 'CLIVT', 'CLIVED')
```
